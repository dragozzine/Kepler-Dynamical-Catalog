```python
import pandas as pd
# Loading file
filename = input("Filename?")
print(f"Loading file {filename}.csv...")
df = pd.read_csv(f"{filename}.csv", low_memory=False)
print(f"{filename}.csv loaded")
```

```python
print(df)
```

```python
# Setting up input loops
check1 = False
check2 = False
allrows = False
confirm2 = False
finish = False
# Setting up variables to be used in partitioning rows and columns
if filename == "kmdc_7_13_26":
    koi_name = "kmdc_index"
    df_wout_col = df.drop(koi_name, axis=1)
    col = [koi_name]
    kmdc = df[koi_name].astype(str)
elif filename == "final_kdc":
    koi_name = "KIC"
    df_wout_col = df.drop(koi_name, axis=1)
    col = [koi_name]
    kmdc = df[koi_name].astype(str)
else:
    print("Unknown file. Edit the python script to read this. Exiting as a precaution.")
    raise SystemExit

while not check1:
    inpcol = input(
        "What column(s) would you like? Enter the names EXACTLY as they appear in the files.\n"
        f"Write the column names with ',' between them. The {koi_name} column will be included automatically.\n"
    )
    inpcol = [col.strip() for col in inpcol.split(',')]
    invalid = [x for x in inpcol if x not in df_wout_col.columns]
    # These "if else" statements are to ensure that the name(s) of the parameter(s) are entered correctly.
    # If it's not entered correctly, the user is prompted again in a loop until they enter the correct name(s) of the parameter(s)
    if invalid:
        print(f"Invalid column(s): {', '.join(invalid)}")
        print(f"Your options are: {', '.join(df_wout_col.columns)}")
    else:
        # Creating a confirmation loop to ensure the user has the correct columns
        conf = False
        while not conf:
            confirm = input(f"Your columns are: {col+inpcol}. Confirm? (y/n)").lower()
            if confirm in ('y', 'yes'):
                conf = True
            elif confirm in ('n', 'no'):
                conf = True
            else:
                print("Please type 'y' or 'n' to confirm.")
        if conf == True and confirm in ('y', 'yes'):
            col.extend(inpcol)
            check1 = True
        else:
            continue
# Creating loop to partition KOI (if desired by the user)
while not confirm2:
    koi_check = input("Do you want every KOI? (y/n): ").lower()
    if koi_check in ('y', 'yes'):
        allrows = True
        confirm2 = True
    elif koi_check in ('n', 'no'):
        valid_kois = set(df[koi_name].astype(str))
        while not check2:
            # Here we begin a similar process to the check1 loop
            inpkois = input("Enter the KOI number(s) (e.g. 456, 992, 1824): ")
            kois = [koi.strip() for koi in inpkois.split(",")]
            valid = []
            invalid = []
            for koi in kois:
                if kmdc.str.startswith(koi).any():
                    valid.append(koi)
                else:
                    invalid.append(koi)
            if invalid:
                print("No KOIs found beginning with:")
                print(", ".join(invalid))
            else:
                matches = df[kmdc.str.startswith(tuple(valid))]
                print("Columns:", col)
                print("KOIs:", ", ".join(valid))
                check2 = True
                confirm2 = True
    else:
        print("Please type 'y' or 'n' to confirm.")

# Final confirmation before creating file
while not finish:
    print()
    print("We will make a .csv file with:")
    print("Columns:",col)
    if allrows:
        print("KOI: All")
    else:
        print("KOIs:", ", ".join(valid))
    final_confirm = input("Final confirm? (y/n)\n")
    if final_confirm in ('y', 'yes'):
        if allrows:
            output_df = df[col]
            finish = True
        else:
            output_df = matches[col]
            finish = True
    elif final_confirm in ('n', 'no'):
        print("No file was created. Please rerun the script to change your selections.")
        raise SystemExit
    else:
        print("Please type 'y' or 'n' to confirm.")

# Creating the file
filename = input("What would you like to call the output file? ")
output_df.to_csv(f"{filename}.csv")
print(f"Saved {filename}.csv")
```
